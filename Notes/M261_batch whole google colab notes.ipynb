Slicing:

It is a phenomenon of extracting group of values from a collection sequentially.

Syntax:

var_name[SI:EI+-1:update/steps]

Simplified Syntaxes:

if SI==0 or SI==-1:

var_name[:EI+-1:steps]

if EI== len(collection):
var_name[SI::steps]

if steps==1:

var_name[SI:EI+-1]

reverse of a whole string: var_name[::-1]

#Copy Operation

It is a process of copying the content from one varible to another keeping the original content intact.

There are 2 types:

General/Normal copy
1. Shallow copy
2. Deep copy

int,float,complex we cannot apply copy operation cz it is immutable.

str,tuple--immutable

set--no duplicate

dict-- key-value is to be stored

Only compatiable datatype is list[mutable datatype],store duplicate values.

#General/Normal Copy

The content of the variable will be copied in the same memory location.

#Shallow Copy

The content of the variable is copied in different memory location.

Syntax:

import copy
copy.copy(source_var)

or

dest_var=source_var.copy()


In nested collection when shallow copy is done the nested values gets copied in the same memory location which when modified can effect the original collection.

#Deep Copy

It copies the content of the original variable into another variable but in different memory location, even if it is a nested collection.

Syntax:

import copy
dest_var=copy.deepcopy(source_var)

The library function which are meant to perform some specific task or operation.

Eg:  10+20

Types:

1. Arithmetic (+,-,/,*,....)
2. Logical (AND,OR,NOT)
3. Relational (<,>,....)
4. Bitwise (&,|,>>,<<,.....)
5. Assignment (=)
6. Membership (in,not in)
7. Identity (is, is not)

1. Arithmetic Operator: are used to perform mathematical calculations

i. Addition(+):  op1+op2 [all SVD,str,list,tuple supports]
ii. Substraction(-): op1-op2 [all SVD,set supports]
iii. Multiplication(*): op1*op2 [SVD]
                        op*n [MVD] n:no of times the value should be multiplied.value of n should be always integer.
[all SVD, string,list, tuple supports]
iv.Division:
    . True Division (/): gives all the values including the float number.
    Eg: 10/3=3.3333333333333335
    . Floor Division(//): gives only the real number.
    Eg: 10//3=3

#Global and Local Variable

Global variable are the variable which is defined outside the function, these variables are aceessed anywhere in the program.

to modify any global variable inside th function we need to have the permission to modify it, to modify we have to use the keyword 'global'.

Local variable are the variable which is declared inside the function, and it can be accessed only inside the function.

to modify local var inside any nested func we have to use 'nonlocal' keyword.

#Difference between gloabl and local variable.

Global var:
.created in main program space.

.scope of the global var is both inside and outside the func.

.to modify global var inside func we have to use 'global' keyword.

Local var:

.created inside the function.

.scope of the var is only inside it's parent func.

.to modify the local var inside nested func we use 'nonlocal' keyword.

```python
#global var:

a=1000
b=2000
def demo():
  global a
  global b
  print(f'The value of a:{a} and b:{b} inside the func.')
  a=1234
  b=2345
  print(f'Modified value of a:{a} and b:{b}.')
demo()
b=7890
print(f'The value of a:{a} and b:{b} outside the func.')

def abcd():
  print(a)
abcd()
```

```python
#local var:

def abc():
  x=1000
  y=2000
  print(f'The value of x: {x} and y:{y}.')
abc()
```

```python
#local var in nested func:

def outer():
  m=1000
  n=2000
  print(f'The value of m:{m} & n:{n} inside outer func.')

  def inner():
    nonlocal m
    nonlocal n
    print(f'The value of m:{m} & n:{n} inside inner func.')
    m=9876
    n=5672
    print(f'Modified value of m:{m} & n:{n} inside inner func')
  inner()
outer()
```

#Packing and Unpacking

It is a process of wrapping up of multiple values in a single variable.

There are 2 types:

1. Single/Tuple packing
2. Double/ Dictionary packing

1. Single/tuple packing:where we can group individual data item in the form of tuple using the packing concept.

By default all the values will be packed in a single var in the tuple format, as it is the most secured datatype.

2. Double/dictionary packing: It is used to wrap key-value pair.

#Types of Methods

There are 3 types of methods:

1. object method: these are used to modify or access the object properties. it is compulsary to store the first parameter as the 'self'.


2. class method: these are used to access or modify the class properties. to access or modify the class properties we have to use 'cls' as the first parameter, @classmethod [decorator].

3. static method: it is neither related to class members or object members but will act as a supportive method.
to create static method we have to use @staticmethod [decorator].

```python
class Company:
  #class properties
  c_name='TCS'
  c_loc='Bangalore'
  login_hour='9hours'

  #constructor method with object properties

  def __init__(self,name,eid,contact,email,job,salary):
    self.name=name
    self.eid=eid
    self.contact=contact
    self.email=email
    self.job=job
    self.salary=salary
#access of object properties
  def display_details(self):
    print(self.name,self.eid,self.salary,self.job,self.email,self.contact)
#modify object properties
  def ch_salary(self,new):
    self.salary=new
    print(self.salary)

#class method to access the class property
  @classmethod
  def display(cls):
    print(cls.c_name,cls.c_loc,cls.login_hour)

#class method to modify the class property
  @classmethod
  def ch_location(cls,new):
    cls.c_loc=new
    print(cls.c_loc)
#static method
  @staticmethod
  def annual_salary_calc(monthly_salary):
    annual_salary=monthly_salary*12
    return annual_salary

#object creation

E1=Company('Atul','TCS1001',9876543210,'atul@tcs.in','Developer',80000)
E2=Company('Varshini','TCS1123',9876543211,'varshini@tcs.in','Tester',85000)
E3=Company('Brindha','TCS1234',9876543210,'brindha@tcs.in','Analyst',90000)


#Object Method
#E1.display_details()
#E2.display_details()
#E3.display_details()

#E2.ch_salary(95000)
#E2.display_details()


#Class Method

#E1.display()
#E3.display()
#E3.ch_location('Kolkata')


#Static Method

E3.annual_salary_calc(90000)


E3.display_details()
```

```python
class Circle:

    # Class property
    r= 5

    # Object method
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        area = 3.14 * self.radius * self.radius
        return area

    # Class method
    @classmethod
    def change_default_radius(cls, new_radius):
        cls.r = new_radius

    # Static method
    @staticmethod
    def get_pi():
        return 3.14


# Creating objects
c1 = Circle(10)
c2 = Circle(7)

# Calling object method
print("Area of Circle 1:", c1.calculate_area())
print("Area of Circle 2:", c2.calculate_area())

# Calling class method
Circle.change_default_radius(10)
print("New Default Radius:", Circle.r)

# Calling static method
print("Value of Pi:", Circle.get_pi())
```

#Encapsulation

It is a process of protecting the data(object memeber) inside the class.

Encapsulation is also known as access specifier.

Based on the term access specifier it can be categoried into following categories:

1. public access specifier: it is a type where we can modify or access the object memebers anywhere in the class.

Syntax:  var=value

2. protected access specifier: it is a type where the object memebrs can be modified/ accessed only inside the class/subcalss.

Syntax: _var=value  (single underscore)

3. private access specifier: it is a type where the object memebers can be accessed/modified only inside the respective class.

Syntax: __var=value (double underscore)

```python
#public access specifier:

class Student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks  #public access specifier

  def display(self):
    print(f'Student Name:{self.name} , Percentage: {self.marks}.')

  def ch_marks(self,new_marks):
    self.marks=new_marks


#object creation

s1= Student('Ishani',91)
s1.display()
s1.ch_marks(95)
s1.display()
```

```python
#protected access specifier:

class Student:
  def __init__(self,name,marks):
    self.name=name
    self._marks=marks  #protected access specifier

  def display(self):
    print(f'Student Name:{self.name} , Percentage: {self._marks}.')

class College(Student):
  def show(self):
     print(f'Student Name:{self.name} , Percentage: {self._marks}.')

#object creation

s2=College('Rohan',82)

s2.display()

print(s2._marks) #access outside the class...recommended not to access outside to avoid data loss.

s2.show()
```

```python
#property decorator
'''
1. it is compulsary to use getter and setter method.

2. @property [decorator] should be used with the method which will return the value. [accessing the object member]

3. both the method name should be same with the var name.

4. for method which we will used to modify values we have to use @var.setter [decorator].
'''


#code

class Student:
  def __init__(self,marks):
    self.__marks=marks

  @property
  def marks(self): #access
    return self.__marks


  @marks.setter
  def marks(self,new_marks): #modify
    if new_marks>0:
      self.__marks=new_marks
    else:
      print('invalid')

#object creation

s5=Student(78)
#print(s5.__marks)

print(s5.marks)

s5.marks +=5
print(s5.marks)
```

#DECORATORS

It is a function which adds special/additional feature in the program.

```python
#Syntax of user defined decorator:

def decorator_name(func):  #func stores the address of the main func
  def inner(*args,**kwargs): #packing values
    #pretask
    func(*args,*kwargs) #main func calling ; unpacking values
    #posttask
  return inner

@decorator_name
def f_name(args):
  #SB
f_name(parameter)
```

```python
#A--B--C

print('visit website www.instagram.com')#pre task
print('login to account')#pre task
print('post a photo') #main task
print('logout from the account')#post task

print()

print('visit website www.instagram.com')#pre task
print('login to account') #pre task
print('like the post') #main task
print('logout from the account')#post task

print()

print('visit website www.instagram.com')#pre task
print('login to account')#pre task
print('comment on the post') #main task
print('logout from the account')#post task
```

```python
def instagram(func):
  def inner(*a,**b):
    print('visit website www.instagram.com')#pre task
    print('login to account')

    func(*a,*b)  #main func calling

    print('logout from the account') #post task
  return inner

@instagram
def A():
  print('post a photo')
A()

print()

@instagram
def B():
  print('like the photo')
B()

print()

@instagram
def C():
  print('comment on the photo')
C()
```

```python
#Create a decorator to find the total execution time of a program.

import time

def timer(func):
  def inner(*a,**b):
    start_timer=time.time() #pre task
    func(*a,*b) #main func calling
    end_timer=time.time() #post task
    print('Total execution: ',end_timer-start_timer)#post task

  return inner

@timer
def add(n,m):
  print('Sum:',n+m)
add(45,45)

@timer
def rev(n):
  print('Reversed string:',n[::-1])
rev('data')
```

```python
#Create a decorator which will provide 5 secs delay of program before execution.

import time

def delay(func):
  def inner(*a,**b):
    time.sleep(7) #pre task
    func(*a,*b)
  return inner

@delay
def multiply(a,b):
  print('Product of: ',a*b)
multiply(8,9)
```

```python
#how to handle the exceptions?

#to handle exceptions there are 4 blocks:

try:
  #code with the error
except Exception_name:
  #soltuion of the error
else:  #optional
  #this block will execute if there is no excpetion/error
finally:
  #this part gets executed everytime.
```

```python
try:
  a=int(input('Enter a:'))
  b=int(input('Enter b: '))
  div=a/b
  print(div)
except (ZeroDivisionError,ValueError) as d:
  print(d)
else:
  for i in range(1,11):
    print(i,end=',')
finally:
  print('\nhello world')
```

```python
#user defined exception:

#Steps:

#create a exception class from the base class i.e: Exception.
#use the keyword 'raise' to raise the exception.
#give the exception a solution.

#Create an exception where a integer no cannot be divided with 6.

class SixDivisionError(Exception):
  pass

try:
  n1=int(input('Enter integer no:'))
  n2=int(input('Enter integer no:'))
  if n2==6:
    raise SixDivisionError
  div=n1/n2
  print(div)

except SixDivisionError:
  print('division by 6 is not possible.')

except ValueError as e:
  print(e)
```

#Regular Expression (RegEx)


It is aprocess where by writing an expression any match/pattern will be fetched.

Tokens:

[a-z]: to match lowercase char
[0-9] or \d: to match digits
[A-Za-z0-9] or \w: to match alphanumeric char.

\: it is used to match any pattern with special char.

\#  \*

use the module 're'

```python
#search(): re.search(pattern,data) it will give us the output for the first matched pattern from the data.

import re

text='Python is a high-level programming language. Python is easy to read and write programming language. Python is a multi-paradigm programming language.'

pattern='python'

s=re.search(pattern,text,re.IGNORECASE)

print(s)
```

```python
#match(): re.match(pattern,data)  it will give us the very first matched pattern from the data.

import re

text='Python is a high-level programming language. Python is easy to read and write programming language. Python is a multi-paradigm programming language.'

pattern='python'
m=re.match(pattern,text,re.IGNORECASE)
print(m)

pattern1='language'
n=re.match(pattern1,text)
print(n)
```

```python
#finditer(): re.finditer(pattern,data) it gives all the matched patterns from the data.

import re

text='Python is a high-level programming language. Python is easy to read and write programming language. Python is a multi-paradigm programming language.'

pattern='is'

iter=re.finditer(pattern,text)
for i in iter:
  print(i)
```

```python
#findall(): re.findall(pattern,data) it will give all the matched patterns in the form of list.

import re

text='Python is a high-level 9 programming language. Python is easy to read and 9 write programming language.9 Python is a multi-paradigm programming 9 language.'

pattern='9'
find=re.findall(pattern,text)
print(find)
```
